```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import math

# -----------------------------
# Input Data
# -----------------------------
attendance = np.array([92, 85, 78, 60, 95, 88, 72, 58, 90, 80,
                       65, 55, 98, 82, 75, 68, 93, 70, 85, 62])

marks = np.array([88, 81, 72, 55, 91, 84, 68, 50, 87, 74,
                  60, 48, 94, 77, 70, 62, 89, 65, 80, 57])

students = np.arange(1, 21)

# -----------------------------
# (a) Bar Chart
# -----------------------------
plt.figure(figsize=(12, 6))
width = 0.4

plt.bar(students - width/2, attendance,
        width=width, label='Attendance (%)')

plt.bar(students + width/2, marks,
        width=width, label='Final Marks')

plt.xlabel('Student Number')
plt.ylabel('Percentage / Marks')
plt.title('Student Attendance and Final Marks')
plt.xticks(students)
plt.legend()
plt.show()

# -----------------------------
# (b) Scatter Plot
# -----------------------------
plt.figure(figsize=(8, 6))
plt.scatter(attendance, marks)
plt.xlabel('Attendance (%)')
plt.ylabel('Final Marks')
plt.title('Attendance vs Final Marks')
plt.grid(True)
plt.show()

# -----------------------------
# (c) Linear Regression
# -----------------------------
X = attendance.reshape(-1, 1)
y = marks

model = LinearRegression()
model.fit(X, y)

# Predicted Marks
y_pred = model.predict(X)

# Regression Equation
slope = model.coef_[0]
intercept = model.intercept_

print("Regression Equation:")
print(f"Marks = {intercept:.2f} + {slope:.2f} × Attendance")

# MSE and RMSE
mse = mean_squared_error(y, y_pred)
rmse = math.sqrt(mse)

print("\nMean Squared Error (MSE):", round(mse, 2))
print("Root Mean Squared Error (RMSE):", round(rmse, 2))

# Additional Metrics (matches your expected values)
mae = mean_absolute_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("Mean Absolute Error (MAE):", round(mae, 2))
print("R² Score:", round(r2, 3))

# -----------------------------
# Regression Line Plot
# -----------------------------
plt.figure(figsize=(8, 6))
plt.scatter(attendance, marks, label='Actual Data')
plt.plot(attendance, y_pred,
         linewidth=2,
         label='Regression Line')

plt.xlabel('Attendance (%)')
plt.ylabel('Final Marks')
plt.title('Regression Line for Marks Prediction')
plt.legend()
plt.grid(True)
plt.show()
```
